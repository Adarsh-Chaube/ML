```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
"""Loading the DATA"""
df=pd.read_csv(r"C:\Users\lenovo\Desktop\Projects\dataset\titanic.csv")
```

```python
"""Quick Overview"""
print(df.head())
print(df.info())
print(df.describe)
print(df.isnull().sum())
```

```python
"""Univariate Analysis"""
df['Age'].hist(bins=30)
plt.show()

print(df['Sex'].value_counts())
sns.countplot(data=df,x="Sex")
plt.show()

sns.countplot(data=df, x='Pclass')
plt.show()

sns.countplot(data=df, x='Embarked')
plt.show()
```

```python
"Bivariate Analysis"
sns.countplot(data=df,x="Sex",hue="Survived")
plt.show()

sns.countplot(data=df,x="Pclass",hue="Survived")
plt.show()

sns.histplot(data=df,x="Age",hue="Survived",multiple='stack')
plt.show()

df['FamilySize']=df['SibSp']+df['Parch']+1
sns.countplot(data=df,x='FamilySize',hue='Survived')
```

```python
"""Multivariate Analysis"""
sns.heatmap(df.corr(numeric_only=True),annot=True, cmap='coolwarm')

sns.catplot(data=df, x='Pclass', hue='Survived', col='Sex', kind='count')
```

```python
"""Drop Cabin as it has many missing values"""
df.drop(columns=['Cabin'],inplace=True)
```

```python
"""Dealing with Missing Values"""
df.fillna({'Age': df['Age'].median(),
           'Embarked': df['Embarked'].mode()[0]}, inplace=True)
```
